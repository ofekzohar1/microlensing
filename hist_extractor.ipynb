```python
import pandas as pd


histogram_dict: ValErrDict = {}
for name, val_list in param_list_values.items():
    if name == U_MIN:
        val_list = [val for val in val_list if val > 0.6780]
    histogram_dict[name] = ms_utils.norm_hist(name, val_list)
```

```python
for name, hist in histogram_dict.items():
    print(name, hist.value, hist.upper_error, hist.lower_error)
```
